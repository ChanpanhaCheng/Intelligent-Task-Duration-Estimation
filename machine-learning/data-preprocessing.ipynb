#### Q5. Compute Pearson correlation coefficients between numerical variables. Plot the coefficients on a heatmap.

`lead_time_days` is the dataset’s only numerical measurement. For this exploratory comparison, represent `Created` and `Resolved` as calendar days since the earliest creation date. Lead time is calculated from those dates, so the correlations partly reflect that arithmetic relationship. These results describe this dataset. A model making estimates at issue creation would not have the resolution date.

```python
# created = pd.to_datetime(df["Created"], format="%Y-%m-%d")
# resolved = pd.to_datetime(df["Resolved"], format="%Y-%m-%d")
created = pd.to_datetime(df["Created"], utc=True)
resolved = pd.to_datetime(df["Resolved"], utc=True)
first_created = created.min()

correlation_data = pd.DataFrame({
    "Created (days)": (created - first_created).dt.days,
    "Resolved (days)": (resolved - first_created).dt.days,
    "actual_completion_duration_days (days)": df["actual_completion_duration_days"],
})
correlations = correlation_data.corr(method="pearson")
display(correlations.round(3))

fig, ax = plt.subplots(figsize=(7, 5), layout="constrained")
heatmap = ax.imshow(correlations.to_numpy(), cmap="coolwarm", vmin=-1, vmax=1)
labels = correlations.columns.tolist()
ax.set_xticks(range(len(labels)), labels, rotation=20, ha="right")
ax.set_yticks(range(len(labels)), labels)
ax.set_title("Pearson correlation")
for row in range(len(labels)):
    for col in range(len(labels)):
        value = correlations.iat[row, col]
        ax.text(col, row, f"{value:.2f}", ha="center", va="center",
                color="white" if abs(value) >= 0.7 else "black")
fig.colorbar(heatmap, ax=ax, label="Pearson r")
plt.show()
```

### 2. Data Preprocessing

#### Q1. Find the redundancy records in the dataset

Check for exact copies across all columns, then flag possible duplicates by matching a normalized title and creation date. Verify those matches against Jira if issue keys are available.

```python
exact_match = df.duplicated(keep=False)
print(f"Rows involved in exact duplicates: {exact_match.sum():,}")
display(df.loc[exact_match])

normalized_title = (
    df["title"].astype("string").str.strip().str.replace(r"\s+", " ", regex=True).str.casefold()
)
duplicate_keys = pd.DataFrame({"normalized_title": normalized_title, "Created": created})
valid_key = normalized_title.notna() & normalized_title.ne("") & created.notna()
possible_match = valid_key & duplicate_keys.duplicated(keep=False)

match_columns = ["title", "Created", "Resolved", "Priority", "task_type", "actual_completion_duration_days"]
possible_duplicates = (
    df.loc[possible_match, match_columns]
    .assign(normalized_title=normalized_title[possible_match])
    .sort_values(["normalized_title", "Created"])
    .drop(columns="normalized_title")
)
pair_groups = duplicate_keys.loc[possible_match].drop_duplicates().shape[0]
print(f"Possible duplicate groups: {pair_groups:,} ({len(possible_duplicates):,} rows)")
display(possible_duplicates)
```
